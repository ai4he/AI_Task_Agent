```python
df1 = pd.read_csv('Pre-test+1_August+31,+2024_17.24.csv')
df1.drop([0,1], inplace=True)
df1 = df1[pd.to_datetime(df1['StartDate'], format='%Y-%m-%d %H:%M:%S') > pd.to_datetime('2024-08-23')]
df1.drop(columns=['StartDate','EndDate','Progress','Finished','Status','IPAddress','RecordedDate','ResponseId',
                  'RecipientLastName','RecipientFirstName','RecipientEmail','ExternalReference','LocationLatitude',
                  'LocationLongitude','DistributionChannel','UserLanguage'], axis=1, inplace=True)

applications = ['Gmail', 'Google Calendar', 'Google Drive', 'Google Maps', 'Google Tasks', 'None', 'Other (input)']
for app in applications:
    name = {
        'Gmail': 'pre_usage_gmail',
        'Google Calendar': 'pre_usage_calendar',
        'Google Drive': 'pre_usage_drive',
        'Google Maps': 'pre_usage_maps',
        'Google Tasks': 'pre_usage_tasks',
        'None': 'pre_usage_none',
        'Other (input)': 'pre_usage_other'
    }.get(app)
    df1[name] = df1['Q1'].apply(lambda x: 1 if app in x else 0)
df1.drop(columns=['Q1'], inplace=True)

df1.Q1_6_TEXT.fillna('', inplace=True)
df1['Q1_6_TEXT'] = df1['Q1_6_TEXT'].apply(lambda x: ','.join([v for v in x.split(',') if 'chrome' not in v.lower()]))

# df1['Q2'].replace('1 hour or less', 1, inplace=True)

challenges = ['Time management', 'Coordination with team members', 'Repetitive tasks', 'Prioritization', 'Not facing any challenges', 'Other (input)']
for challenge in challenges:
    name = {
        'Time management': 'pre_challenge_timemanagement',
        'Coordination with team members': 'pre_challenge_coordination',
        'Repetitive tasks': 'pre_challenge_repetitive',
        'Prioritization': 'pre_challenge_prioritization',
        'Not facing any challenges': 'pre_challenge_none',
        'Other (input)': 'pre_challenge_other'
    }.get(challenge)
    df1[name] = df1['Q8'].apply(lambda x: 1 if challenge in x else 0)
df1.drop(columns=['Q8'], inplace=True)

planned_help = ['Automating repetitive tasks', 'Providing task recommendations', 'Assisting in task prioritization', 'Don\'t know', 'Other (input)']
for help in planned_help:
    name = {
        'Automating repetitive tasks': 'pre_help_automating_repetitive_tasks',
        'Providing task recommendations': 'pre_help_providing_task_recommendations',
        'Assisting in task prioritization': 'pre_help_assisting_in_task_prioritization',
        'Don\'t know': 'pre_help_dont_know',
        'Other (input)': 'pre_help_other'
    }.get(help)
    df1[name] = df1['Q14'].apply(lambda x: 1 if help in x else 0)
df1.drop(columns=['Q14'], inplace=True)

benefits = ['Increased productivity', 'Better task organization', 'Time savings', 'Reduced workload', 'Other (input)']
for benefit in benefits:
    name = {
        'Increased productivity': 'pre_benefit_increased_productivity',
        'Better task organization': 'pre_benefit_better_task_organization',
        'Time savings': 'pre_benefit_time_savings',
        'Reduced workload': 'pre_benefit_reduced_workload',
        'Other (input)': 'pre_benefit_other'
    }.get(benefit)
    df1[name] = df1['Q16'].apply(lambda x: 1 if benefit in x else 0)
df1.drop(columns=['Q16'], inplace=True)

foreseen_challenges = ['Learning curve', 'Trust in AI\'s abilities', 'Integration with existing tools/systems', 'Privacy/security concerns', 'Other (input)']
for challenge in foreseen_challenges:
    name = {
        'Learning curve': 'pre_foreseen_challenge_learning_curve',
        'Trust in AI\'s abilities': 'pre_foreseen_challenge_trust_in_AI_abilities',
        'Integration with existing tools/systems': 'pre_foreseen_challenge_integration_with_existing_tools',
        'Privacy/security concerns': 'pre_foreseen_challenge_privacy_security_concerns',
        'Other (input)': 'pre_foreseen_challenge_other'
    }.get(challenge)
    df1[name] = df1['Q20'].apply(lambda x: 1 if challenge in x else 0)
df1.drop(columns=['Q20'], inplace=True)

ordered = ['Q11','Q29','Q30','Q30_8_TEXT','Q31','Duration (in seconds)','pre_usage_gmail','pre_usage_calendar','pre_usage_drive',
           'pre_usage_maps','pre_usage_tasks','pre_usage_none','pre_usage_other','Q1_6_TEXT','Q2','Q4','Q7','Q33','Q34','Q35','Q36',
           'pre_challenge_timemanagement','pre_challenge_coordination','pre_challenge_repetitive','pre_challenge_prioritization',
           'pre_challenge_none','pre_challenge_other','Q8_6_TEXT','Q9','Q37','Q12','pre_help_automating_repetitive_tasks',
           'pre_help_providing_task_recommendations','pre_help_assisting_in_task_prioritization','pre_help_dont_know','pre_help_other',
           'Q14_4_TEXT','Q15','pre_benefit_increased_productivity','pre_benefit_better_task_organization','pre_benefit_time_savings',
           'pre_benefit_reduced_workload','pre_benefit_other','Q16_5_TEXT','Q18','Q19','pre_foreseen_challenge_learning_curve',
           'pre_foreseen_challenge_trust_in_AI_abilities','pre_foreseen_challenge_integration_with_existing_tools',
           'pre_foreseen_challenge_privacy_security_concerns','pre_foreseen_challenge_other','Q20_5_TEXT','Q23_1','Q23_2','Q23_3',
           'Q23_4','Q23_5','Q24','Q26','Q27']
df1 = df1[ordered + [ col for col in df1.columns if col not in ordered]]

df1.rename(columns={
    'Q11': 'prolific_id',
    'Duration (in seconds)': 'pre_duration',
    'Q29': 'gender',
    'Q30': 'ethnicity',
    'Q30_8_TEXT': 'ethnicity_other',
    'Q31': 'age',
    'Q1_6_TEXT': 'pre_usage_other_details',
    'Q2': 'pre_hours_per_day',
    'Q4': 'pre_activities',
    'Q7': 'pre_usage_gmail_easy',
    'Q33': 'pre_usage_calendar_easy',
    'Q34': 'pre_usage_drive_easy',
    'Q35': 'pre_usage_maps_easy',
    'Q36': 'pre_usage_tasks_easy',
    'Q8_6_TEXT': 'pre_challenge_other_details',
    'Q9': 'pre_used_automation_was_convenient',
    'Q37': 'pre_comfortable_with_idea_of_automation',
    'Q12': 'pre_always_use_AI_if_possible',
    'Q14_4_TEXT': 'pre_help_other_details',
    'Q15': 'pre_used_LLM',
    'Q16_5_TEXT': 'pre_benefit_other_details',
    'Q18': 'pre_expectations',
    'Q19': 'pre_used_ai_task_management_agents',
    'Q20_5_TEXT': 'pre_foreseen_challenge_other_details',
    'Q23_1': 'pre_expertise_gmail',
    'Q23_2': 'pre_expertise_calendar',
    'Q23_3': 'pre_expertise_drive',
    'Q23_4': 'pre_expertise_maps',
    'Q23_5': 'pre_expertise_tasks',
    'Q24': 'pre_expertise_chatgpt',
    'Q26': 'pre_used_chatgpt_to_accomplish_tasks',
    'Q27': 'pre_used_chatgpt_purpose'
}, inplace=True)

df1.ethnicity_other.fillna('', inplace=True)
df1.pre_challenge_other_details.fillna('', inplace=True)
df1.pre_used_automation_was_convenient.fillna('', inplace=True)
df1.pre_help_other_details.fillna('', inplace=True)
df1.pre_used_LLM.replace({'Yes': 1, 'No': 0}, inplace=True)
df1.pre_benefit_other_details.fillna('', inplace=True)
df1.pre_used_ai_task_management_agents.replace({'Yes': 1, 'No': 0}, inplace=True)
df1.pre_foreseen_challenge_other_details.fillna('', inplace=True)
df1.pre_used_chatgpt_purpose.fillna('', inplace=True)

df1 = df1.astype({
    'prolific_id': 'string',
    'gender': 'category',
    'ethnicity': 'category',
    'ethnicity_other': 'string',
    'age': 'category',
    'pre_duration': 'int32',
    'pre_usage_gmail': 'bool',
    'pre_usage_calendar': 'bool',
    'pre_usage_drive': 'bool',
    'pre_usage_maps': 'bool',
    'pre_usage_tasks': 'bool',
    'pre_usage_none': 'bool',
    'pre_usage_other': 'bool',
    'pre_usage_other_details': 'string',
    'pre_hours_per_day': 'category',
    'pre_activities': 'string',
    'pre_usage_gmail_easy': 'category',
    'pre_usage_calendar_easy': 'category',
    'pre_usage_drive_easy': 'category',
    'pre_usage_maps_easy': 'category',
    'pre_usage_tasks_easy': 'category',
    'pre_challenge_timemanagement': 'bool',
    'pre_challenge_coordination': 'bool',
    'pre_challenge_repetitive': 'bool',
    'pre_challenge_prioritization': 'bool',
    'pre_challenge_none': 'bool',
    'pre_challenge_other': 'bool',
    'pre_challenge_other_details': 'string',
    'pre_used_automation_was_convenient': 'category',
    'pre_comfortable_with_idea_of_automation': 'category',
    'pre_always_use_AI_if_possible': 'category',
    'pre_help_automating_repetitive_tasks': 'bool',
    'pre_help_providing_task_recommendations': 'bool',
    'pre_help_assisting_in_task_prioritization': 'bool',
    'pre_help_dont_know': 'bool',
    'pre_help_other': 'bool',
    'pre_help_other_details': 'string',
    'pre_used_LLM': 'bool',
    'pre_benefit_increased_productivity': 'bool',
    'pre_benefit_better_task_organization': 'bool',
    'pre_benefit_time_savings': 'bool',
    'pre_benefit_reduced_workload': 'bool',
    'pre_benefit_other': 'bool',
    'pre_benefit_other_details': 'string',
    'pre_expectations': 'string',
    'pre_used_ai_task_management_agents': 'bool',
    'pre_foreseen_challenge_learning_curve': 'bool',
    'pre_foreseen_challenge_trust_in_AI_abilities': 'bool',
    'pre_foreseen_challenge_integration_with_existing_tools': 'bool',
    'pre_foreseen_challenge_privacy_security_concerns': 'bool',
    'pre_foreseen_challenge_other': 'bool',
    'pre_foreseen_challenge_other_details': 'string',
    'pre_expertise_gmail': 'category',
    'pre_expertise_calendar': 'category',
    'pre_expertise_drive': 'category',
    'pre_expertise_maps': 'category',
    'pre_expertise_tasks': 'category',
    'pre_expertise_chatgpt': 'category',
    'pre_used_chatgpt_to_accomplish_tasks': 'category',
    'pre_used_chatgpt_purpose': 'string'
})

df1.head()
```
